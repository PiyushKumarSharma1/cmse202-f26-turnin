You'll work with a simple `Thermostat` class. A thermostat tracks a **current** temperature and a **target** temperature, can have its target changed, can take a **step** (moving the current temperature one degree toward the target), and can report its **status**.

The starter code below is **partly finished**: the `step` method is already written for you, but three other methods still need to be completed.

&#9989; **Question 4.1 (7 points):** Complete the `Thermostat` class in the cell below so that:
- `__init__(self, current_temp, target_temp)` stores the two values as the attributes `self.current_temp` and `self.target_temp`. (The provided `step` method uses these exact attribute names, so be sure to match them.)
- `set_target(self, new_target)` updates the target temperature (`self.target_temp`) based on the value of `new_target`.
- `status(self)` returns `"heating"` if the current temperature is **below** the target, `"cooling"` if it is **above** the target, and `"at target"` if the two are **equal**.

You do **not** need to write `step` -- it is provided for you (leave it as-is for now; you'll come back to it in Question 4.4).

````python
# Complete __init__, set_target, and status; leave the provided step unchanged here.
class Thermostat:
    def __init__(self, current_temp, target_temp):
        self.current_temp = current_temp
        self.target_temp = target_temp

    def set_target(self, new_target):
        self.target_temp = new_target

    def step(self):
        # provided for you: nudge the current temperature toward the target
        if self.current_temp != self.target_temp:
            self.current_temp += 1

    def status(self):
        if self.current_temp < self.target_temp:
            return "heating"
        elif self.current_temp > self.target_temp:
            return "cooling"
        else:
            return "at target"
````

&#9989; **Question 4.4 (3 points):** The `step` method you were given works when the thermostat needs to **heat up**, but it has a bug (which you might have noticed when testing the class in the previous question). Run the cell below: the thermostat starts *above* its target (current 75, target 70), so it should **cool down** toward 70 -- but the current temperature moves the wrong way.

1. Describe what is wrong with the behavior. *(1 pt)*
2. Fix the bug: in the answer cell provided, write a corrected version of the `step` method (or the whole class) so the thermostat moves one degree toward the target in **both** directions. *(1 pt)*
3. In one sentence, explain what caused the bug. *(1 pt)*

````python
# Do not modify this cell -- it runs the class you built above on a COOLING scenario.
# Starts ABOVE the target, so it should cool DOWN toward 70:
t = Thermostat(75, 70)
for _ in range(3):
    t.step()
print("after 3 steps, current_temp =", t.current_temp, "(target is 70)")
````

````python
# Corrected class: the new step handles heating, cooling, and equality.
class Thermostat:
    def __init__(self, current_temp, target_temp):
        self.current_temp = current_temp
        self.target_temp = target_temp

    def set_target(self, new_target):
        self.target_temp = new_target

    def step(self):
        if self.current_temp < self.target_temp:
            self.current_temp += 1
        elif self.current_temp > self.target_temp:
            self.current_temp -= 1

    def status(self):
        if self.current_temp < self.target_temp:
            return "heating"
        elif self.current_temp > self.target_temp:
            return "cooling"
        else:
            return "at target"

# Recreate the object so it uses the corrected class.
t = Thermostat(75, 70)
for _ in range(3):
    t.step()
print("Corrected after 3 steps:", t.current_temp, t.status())
assert t.current_temp == 72 and t.status() == "cooling"
for _ in range(2):
    t.step()
assert t.current_temp == 70 and t.status() == "at target"
t.step()
assert t.current_temp == 70
````

1. The provided method raises the temperature from 75 to 78 after three steps, moving away from 70 instead of cooling to 72.
3. The bug is caused by adding 1 whenever the temperatures differ, without checking whether the current temperature is above or below the target.
